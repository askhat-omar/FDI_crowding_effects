```python
import pandas as pd
import numpy as np
import camelot
from functools import reduce
from pathlib import Path
```

# Abay oblast and East-Kazakhstan oblast

Since these two oblasts were divided in 2022, we will need to recombine rayon level data for both.

## Investments

### 2022-2024

```python
abay_all_2022 = pd.read_excel(Path("../Data/Investments/Abay/2022_Investments.xlsx"), sheet_name='13', header=6, usecols='A, C, E', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
abay_all_2023 = pd.read_excel(Path("../Data/Investments/Abay/2023_Investments.xlsx"), sheet_name='19', header=3, usecols='A, C, F', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
abay_all_2024 = pd.read_excel(Path("../Data/Investments/Abay/2024_Investments.xlsx"), sheet_name='14', header=3, usecols='A, C, E', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
```

```python
abay_non_2022 = pd.read_excel(Path("../Data/Investments/Abay/2022_Investments.xlsx"), sheet_name='14', header=6, usecols='A, B, D', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
abay_non_2023 = pd.read_excel(Path("../Data/Investments/Abay/2023_Investments.xlsx"), sheet_name='20', header=3, usecols='A, B, D', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
abay_non_2024 = pd.read_excel(Path("../Data/Investments/Abay/2024_Investments.xlsx"), sheet_name='15', header=3, usecols='A, B, D', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
```

```python
all_columns = ['rayon', 'l_all_inv', 'f_all_inv']
non_columns = ['rayon', 'all_nonextr_inv', 'f_nonextr_inv']

for df in [abay_all_2022, abay_all_2023, abay_all_2024]:
    df.columns = all_columns

for df in [abay_non_2022, abay_non_2023, abay_non_2024]:
    df.columns = non_columns
    df['l_nonextr_inv'] = df['all_nonextr_inv'] - df['f_nonextr_inv']
    df.drop('all_nonextr_inv', axis=1, inplace=True)
```

```python
abay_all_2024.loc[0, ['l_all_inv', 'f_all_inv']] = abay_all_2024.loc[0, ['l_all_inv', 'f_all_inv']] + abay_all_2024.loc[7, ['l_all_inv', 'f_all_inv']]
abay_all_2024.loc[11, ['l_all_inv', 'f_all_inv']] = abay_all_2024.loc[10, ['l_all_inv', 'f_all_inv']] + abay_all_2024.loc[11, ['l_all_inv', 'f_all_inv']]
abay_all_2024 = abay_all_2024.drop([7, 10]).reset_index(drop=True)

abay_non_2024.loc[0, ['l_nonextr_inv', 'f_nonextr_inv']] = abay_non_2024.loc[0, ['l_nonextr_inv', 'f_nonextr_inv']] + abay_non_2024.loc[7, ['l_nonextr_inv', 'f_nonextr_inv']]
abay_non_2024.loc[11, ['l_nonextr_inv', 'f_nonextr_inv']] = abay_non_2024.loc[10, ['l_nonextr_inv', 'f_nonextr_inv']] + abay_non_2024.loc[11, ['l_nonextr_inv', 'f_nonextr_inv']]
abay_non_2024 = abay_non_2024.drop([7, 10]).reset_index(drop=True)
```

```python
rayon_names = ['Semey_city', 'Kurchatov_city', 'Abay', 'Aksuat', 'Ayagoz', 'Beskaragay', 'Boroduliha', 'Zharma', 'Kokpekti', 'Urzhar']

for df in [abay_all_2022, abay_all_2023, abay_all_2024, abay_non_2022, abay_non_2023, abay_non_2024]:
    df['rayon'] = rayon_names
```

```python
abay_2022 = abay_all_2022.merge(abay_non_2022, on='rayon')
abay_2023 = abay_all_2023.merge(abay_non_2023, on='rayon')
abay_2024 = abay_all_2024.merge(abay_non_2024, on='rayon')
```

```python
dfs = [abay_2022, abay_2023, abay_2024]

for i in range(3):
    df = dfs[i]
    df['oblast'] = 'East-Kazakhstan'
    if i == 0:
        df['year'] = 2022
    elif i == 1:
        df['year'] = 2023
    else:
        df['year'] = 2024
```

```python
ek_all_2022 = pd.read_excel(Path("../Data/Investments/East-Kazakhstan/2022_Investments.xlsx"), sheet_name='13', header=6, usecols='A, C, E', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
ek_all_2023 = pd.read_excel(Path("../Data/Investments/East-Kazakhstan/2023_Investments.xlsx"), sheet_name='19', header=4, usecols='A, C, F', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
ek_all_2024 = pd.read_excel(Path("../Data/Investments/East-Kazakhstan/2024_Investments.xlsx"), sheet_name='14', header=4, usecols='A, C, E', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
```

```python
ek_non_2022 = pd.read_excel(Path("../Data/Investments/East-Kazakhstan/2022_Investments.xlsx"), sheet_name='14', header=6, usecols='A, B, D', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
ek_non_2023 = pd.read_excel(Path("../Data/Investments/East-Kazakhstan/2023_Investments.xlsx"), sheet_name='20', header=4, usecols='A, B, D', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
ek_non_2024 = pd.read_excel(Path("../Data/Investments/East-Kazakhstan/2024_Investments.xlsx"), sheet_name='15', header=4, usecols='A, B, D', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
```

```python
all_columns = ['rayon', 'l_all_inv', 'f_all_inv']
non_columns = ['rayon', 'all_nonextr_inv', 'f_nonextr_inv']

for df in [ek_all_2022, ek_all_2023, ek_all_2024]:
    df.columns = all_columns

for df in [ek_non_2022, ek_non_2023, ek_non_2024]:
    df.columns = non_columns
    df['l_nonextr_inv'] = df['all_nonextr_inv'] - df['f_nonextr_inv']
    df.drop('all_nonextr_inv', axis=1, inplace=True)
```

```python
ek_all_2024.loc[5, ['l_all_inv', 'f_all_inv']] = ek_all_2024.loc[5, ['l_all_inv', 'f_all_inv']] + ek_all_2024.loc[7, ['l_all_inv', 'f_all_inv']]
ek_all_2024.loc[6, ['l_all_inv', 'f_all_inv']] = ek_all_2024.loc[6, ['l_all_inv', 'f_all_inv']] + ek_all_2024.loc[11, ['l_all_inv', 'f_all_inv']]
ek_all_2024 = ek_all_2024.drop([7, 11]).reset_index(drop=True)

ek_non_2024.loc[5, ['l_nonextr_inv', 'f_nonextr_inv']] = ek_non_2024.loc[5, ['l_nonextr_inv', 'f_nonextr_inv']] + ek_non_2024.loc[7, ['l_nonextr_inv', 'f_nonextr_inv']]
ek_non_2024.loc[6, ['l_nonextr_inv', 'f_nonextr_inv']] = ek_non_2024.loc[6, ['l_nonextr_inv', 'f_nonextr_inv']] + ek_non_2024.loc[11, ['l_nonextr_inv', 'f_nonextr_inv']]
ek_non_2024 = ek_non_2024.drop([7, 11]).reset_index(drop=True)
```

```python
rayon_names = ['Oskemen_city', 'Ridder_city', 'Glubokoe', 'Zaysan', 'Altay', 'Kurshim', 'Katonkaragay', 'Samar', 'Tarbagatay', 'Ulan', 'Shemonaiha']

for df in [ek_all_2022, ek_all_2023, ek_all_2024, ek_non_2022, ek_non_2023, ek_non_2024]:
    df['rayon'] = rayon_names
```

```python
ek_2022 = ek_all_2022.merge(ek_non_2022, on='rayon')
ek_2023 = ek_all_2023.merge(ek_non_2023, on='rayon')
ek_2024 = ek_all_2024.merge(ek_non_2024, on='rayon')
```

```python
dfs = [ek_2022, ek_2023, ek_2024]

for i in range(3):
    df = dfs[i]
    df['oblast'] = 'East-Kazakhstan'
    if i == 0:
        df['year'] = 2022
    elif i == 1:
        df['year'] = 2023
    else:
        df['year'] = 2024
```

```python
east_2022 = pd.concat([abay_2022, ek_2022], ignore_index=True)
east_2023 = pd.concat([abay_2023, ek_2023], ignore_index=True)
east_2024 = pd.concat([abay_2024, ek_2024], ignore_index=True)
```

```python
east_2022.loc[18, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']] = east_2022.loc[18, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']] + east_2022.loc[3, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']]
east_2022.loc[8, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']] = east_2022.loc[8, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']] + east_2022.loc[17, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']]
east_2022 = east_2022.drop([3, 17]).reset_index(drop=True)

east_2023.loc[18, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']] = east_2023.loc[18, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']] + east_2023.loc[3, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']]
east_2023.loc[8, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']] = east_2023.loc[8, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']] + east_2023.loc[17, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']]
east_2023 = east_2023.drop([3, 17]).reset_index(drop=True)

east_2024.loc[18, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']] = east_2024.loc[18, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']] + east_2024.loc[3, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']]
east_2024.loc[8, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']] = east_2024.loc[8, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']] + east_2024.loc[17, ['l_all_inv', 'f_all_inv', 'f_nonextr_inv', 'l_nonextr_inv']]
east_2024 = east_2024.drop([3, 17]).reset_index(drop=True)
```

### 2018-2021

```python
east_all_2018 = camelot.read_pdf(Path("../Data/Investments/East-Kazakhstan/2018_Investments.pdf"), pages='28', flavor='stream')[0].df
east_nonextr_2018 = camelot.read_pdf(Path("../Data/Investments/East-Kazakhstan/2018_Investments.pdf"), pages='29', flavor='stream')[0].df
```

```python
all_columns = ['rayon', 'l_all_inv', 'f_all_inv']
non_columns = ['rayon', 'all_nonextr_inv', 'f_nonextr_inv']

rayons = ['Oskemen_city', 'Kurchatov_city', 'Ridder_city', 'Semey_city', 'Abay', 'Ayagoz', 'Beskaragay', 'Boroduliha', 'Glubokoe', 'Zharma', 'Zaysan', 'Altay', 'Kokpekti', 'Kurshim', 'Katonkaragay', 'Tarbagatay', 'Ulan', 'Urzhar', 'Shemonaiha']

east_all_2018 = east_all_2018[[0, 2, 5]]
east_all_2018 = east_all_2018.drop([0, 1, 2, 3, 4]).reset_index(drop=True)
east_all_2018.columns = all_columns
east_all_2018['rayon'] = rayons
east_all_2018['l_all_inv'] = east_all_2018['l_all_inv'].str.replace(' ', '').astype(int)
east_all_2018['f_all_inv'] = east_all_2018['f_all_inv'].str.replace(' ', '').astype(int)

east_nonextr_2018 = east_nonextr_2018[[0, 2, 5]]
east_nonextr_2018 = east_nonextr_2018.drop([0, 1, 2, 3, 4]).reset_index(drop=True)
east_nonextr_2018.columns = non_columns
east_nonextr_2018['all_nonextr_inv'] = east_nonextr_2018['all_nonextr_inv'].str.replace(' ', '').astype(int)
east_nonextr_2018['f_nonextr_inv'] = east_nonextr_2018['f_nonextr_inv'].str.replace(' ', '').astype(int)
east_nonextr_2018['l_nonextr_inv'] = east_nonextr_2018['all_nonextr_inv'] - east_nonextr_2018['f_nonextr_inv']
east_nonextr_2018 = east_nonextr_2018.drop('all_nonextr_inv', axis=1)
east_nonextr_2018['rayon'] = rayons
```

```python
east_2018 = east_all_2018.merge(east_nonextr_2018, on='rayon')
east_2018['oblast'] = 'East-Kazakhstan'
east_2018['year'] = 2018
```

```python
east_all_2019 = pd.read_excel(Path("../Data/Investments/East-Kazakhstan/2019_Investments.xls"), sheet_name='14', header=6, usecols='A, C, E', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
east_nonextr_2019 = pd.read_excel(Path("../Data/Investments/East-Kazakhstan/2019_Investments.xls"), sheet_name='15', header=6, usecols='A, B, D', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop([0, 20, 21, 22]).reset_index(drop=True)

east_all_2020 = pd.read_excel(Path("../Data/Investments/East-Kazakhstan/2020_Investments.xlsx"), sheet_name='13', header=6, usecols='A, C, E', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
east_nonextr_2020 = pd.read_excel(Path("../Data/Investments/East-Kazakhstan/2020_Investments.xlsx"), sheet_name='14', header=6, usecols='A, B, D', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)

east_all_2021 = pd.read_excel(Path("../Data/Investments/East-Kazakhstan/2021_Investments.xlsx"), sheet_name='13', header=6, usecols='A, C, E', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
east_nonextr_2021 = pd.read_excel(Path("../Data/Investments/East-Kazakhstan/2021_Investments.xlsx"), sheet_name='14', header=6, usecols='A, B, D', na_values=['x', 'х', '-', '', ' ']).fillna(0).drop(0).reset_index(drop=True)
```

```python
rayons = ['Oskemen_city', 'Kurchatov_city', 'Ridder_city', 'Semey_city', 'Abay', 'Ayagoz', 'Beskaragay', 'Boroduliha', 'Glubokoe', 'Zharma', 'Zaysan', 'Altay', 'Kokpekti', 'Kurshim', 'Katonkaragay', 'Tarbagatay', 'Ulan', 'Urzhar', 'Shemonaiha']

all_columns = ['rayon', 'l_all_inv', 'f_all_inv']
non_columns = ['rayon', 'all_nonextr_inv', 'f_nonextr_inv']

for df in [east_all_2019, east_all_2020, east_all_2021]:
    df.columns = all_columns
    df['rayon'] = rayons

for df in [east_nonextr_2019, east_nonextr_2020, east_nonextr_2021]:
    df.columns = non_columns
    df['rayon'] = rayons
    df['l_nonextr_inv'] = df['all_nonextr_inv'] - df['f_nonextr_inv']
    df.drop('all_nonextr_inv', axis=1, inplace=True)
```

## Employment

### 2022-2024

```python
abay_empl = pd.read_excel(Path("../Data/Employment/Abay.xlsx"), skiprows=22, nrows=10, header=None, usecols=[0, 1, 2, 3])
```

```python
abay_empl.columns = ['rayon', 2022, 2023, 2024]

rayons = ['Semey_city', 'Kurchatov_city', 'Abay', 'Aksuat', 'Ayagoz', 'Beskaragay', 'Boroduliha', 'Zharma', 'Kokpekti', 'Urzhar']

abay_empl['rayon'] = rayons

abay_empl[2022] = abay_empl[2022].str.replace(',', '.').astype(float)
abay_empl[2023] = abay_empl[2023].astype(float)
abay_empl[2024] = abay_empl[2024].astype(float)
```

```python
ek_empl = pd.read_excel(Path("../Data/Employment/East-Kazakhstan.xlsx"), skiprows=30, nrows=20, header=None, usecols='A, J, K, L', na_values=['x', 'х', '-', '', ' ']).dropna().reset_index(drop=True)
```

```python
ek_empl.columns = ['rayon', 2022, 2023, 2024]

rayons = ['Oskemen_city', 'Ridder_city', 'Glubokoe', 'Zaysan', 'Altay', 'Samar', 'Kurshim', 'Katonkaragay', 'Tarbagatay', 'Ulan', 'Shemonaiha']

ek_empl['rayon'] = rayons

ek_empl[2022] = ek_empl[2022].astype(float)
ek_empl[2023] = ek_empl[2023].astype(float)
ek_empl[2024] = ek_empl[2024].astype(float)
```

```python
east_empl = pd.concat([abay_empl, ek_empl], ignore_index=True)
```

```python
east_empl.loc[8, [2022, 2023, 2024]] = east_empl.loc[8, [2022, 2023, 2024]] + east_empl.loc[15, [2022, 2023, 2024]]
east_empl.loc[18, [2022, 2023, 2024]] = east_empl.loc[18, [2022, 2023, 2024]] + east_empl.loc[3, [2022, 2023, 2024]]
east_empl = east_empl.drop([3, 15]).reset_index(drop=True)
```

```python
east_empl_22_24 = east_empl.melt(id_vars='rayon', value_vars=[2022, 2023, 2024], var_name='year', value_name='employment')
```

## Other controls

```python
kurchatov = pd.read_excel(Path("../Data/Controls/Abay_All.xls"), sheet_name="моног.Курчатов", header=2, usecols='B, AD:AJ', nrows=272, na_values=['x', 'х', '-', '', ' ', '...'])
```

```python
kurchatov.columns = ['Indicator', 2018, 2019, 2020, 2021, 2022, 2023, 2024]
kurchatov = kurchatov.loc[[134, 167, 257]].reset_index(drop=True)
```

```python
kurchatov['Indicator'] = ['production', 'agriculture', 'retail']
kurchatov[2024] = kurchatov[2024].str.replace(',', '.')
kurchatov[2024] = kurchatov[2024].str.replace(')', '')
kurchatov[2024] = kurchatov[2024].str.replace(' ', '')
kurchatov[2024] = kurchatov[2024].astype(float)
```

```python
kurchatov = pd.melt(kurchatov, id_vars=['Indicator'], var_name='year', value_name='value')
kurchatov = kurchatov.pivot(index='year', columns='Indicator', values='value').reset_index()
kurchatov.insert(0, 'rayon', 'Kurchatov_city')
kurchatov.columns.name = None
```

```python
semey = pd.read_excel(Path("../Data/Controls/Abay_All.xls"), sheet_name="Семей", header=2, usecols='A, AC:AI', nrows=229, na_values=['x', 'х', '-', '', ' ', '...'])
```

```python
semey.columns = ['Indicator', 2018, 2019, 2020, 2021, 2022, 2023, 2024]
semey = semey.loc[[117, 149, 218]].reset_index(drop=True)
```

```python
semey['Indicator'] = ['production', 'agriculture', 'retail']
semey[2024] = semey[2024].str.replace(',', '.')
semey[2024] = semey[2024].str.replace(')', '')
semey[2024] = semey[2024].str.replace(' ', '')
semey[2024] = semey[2024].astype(float)
```

```python
zh_semey = pd.read_excel(Path("../Data/Controls/Abay_All.xls"), sheet_name="Жаңасемей", header=4, usecols='A:B', nrows=229, na_values=['x', 'х', '-', '', ' ', '...'])
```

```python
zh_semey.columns = ['Indicator', 2024]
zh_semey = zh_semey.loc[[117, 149, 218]].reset_index(drop=True)
```

```python
zh_semey['Indicator'] = ['production', 'agriculture', 'retail']
zh_semey[2024] = zh_semey[2024].replace(zh_semey[2024].iloc[2], '1.8')
zh_semey[2024] = zh_semey[2024].str.replace(' ', '')
zh_semey[2024] = zh_semey[2024].str.replace(',', '.')
zh_semey[2024] = zh_semey[2024].str.replace(')', '')
zh_semey[2024] = zh_semey[2024].astype(float)
```

## Investments

### 2022-2024

```python
all_columns = ['rayon', 'l_all_inv', 'f_all_inv']
non_columns = ['rayon', 'all_nonextr_inv', 'f_nonextr_inv']

for df in [zhetisu_all_2022, zhetisu_all_2023, zhetisu_all_2024]:
    df.columns = all_columns

for df in [zhetisu_non_2022, zhetisu_non_2023, zhetisu_non_2024]:
    df.columns = non_columns
    df['l_nonextr_inv'] = df['all_nonextr_inv'] - df['f_nonextr_inv']
    df.drop('all_nonextr_inv', axis=1, inplace=True)

rayon_names = ['Taldykorgan_city', 'Tekeli_city', 'Aksu', 'Alakol', 'Yeskeldi', 'Kerbulak', 'Koksu', 'Karatal', 'Panfilov', 'Sarkan']

for df in [zhetisu_all_2022, zhetisu_all_2023, zhetisu_all_2024, zhetisu_non_2022, zhetisu_non_2023, zhetisu_non_2024]:
    df['rayon'] = rayon_names

zhetisu_2022 = zhetisu_all_2022.merge(zhetisu_non_2022, on='rayon')
zhetisu_2023 = zhetisu_all_2023.merge(zhetisu_non_2023, on='rayon')
zhetisu_2024 = zhetisu_all_2024.merge(zhetisu_non_2024, on='rayon')

dfs = [zhetisu_2022, zhetisu_2023, zhetisu_2024]

for i in range(3):
    df = dfs[i]
    df['oblast'] = 'Almaty'
    if i == 0:
        df['year'] = 2022
    elif i == 1:
        df['year'] = 2023
    else:
        df['year'] = 2024
```

## PPI

```python
almaty_ppi = pd.read_excel(Path("../Data/PPI/Almaty.xls"), header=5, usecols='Z:AF', nrows=238, na_values=['x', 'х', '-', '', ' ', '...'])

almaty_ppi.columns = [2016, 2017, 2018, 2019, 2020, 2021, 2022]

rayons = list(almaty_inv['rayon'].unique())

dfs = []

for rayon in rayons:
    df_temp = pd.DataFrame({
        'rayon': rayon,
        'year': range(2016, 2023),
        'ppi': almaty_ppi.values[0]
    })
    dfs.append(df_temp)

almaty_ppi = pd.concat(dfs, ignore_index=True)

almaty_ppi
```

## Combine together

```python
dfs = [almaty_inv, almaty_empl, almaty_controls, almaty_ppi]

almaty_merged = reduce(lambda left, right: pd.merge(left, right, on=['rayon', 'year'], how='outer'), dfs)

almaty_merged
```

## PPI

```python
aktobe_ppi = pd.read_excel(Path("../Data/PPI/Aktobe.xlsx"), header=2, usecols='AA:AH', nrows=216, na_values=['x', 'х', '-', '', ' ', '...'])

aktobe_ppi.columns = [2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]

rayons = list(aktobe_inv['rayon'].unique())

dfs = []

for rayon in rayons:
    df_temp = pd.DataFrame({
        'rayon': rayon,
        'year': range(2016, 2024),
        'ppi': aktobe_ppi.values[64]
    })
    dfs.append(df_temp)

aktobe_ppi = pd.concat(dfs, ignore_index=True)

aktobe_ppi['ppi'] = aktobe_ppi['ppi'].astype(str).str.replace(',', '.')
aktobe_ppi['ppi'] = aktobe_ppi['ppi'].astype(str).str.replace(')', '')
aktobe_ppi['ppi'] = aktobe_ppi['ppi'].astype(str).str.replace(' ', '')
aktobe_ppi['ppi'] = aktobe_ppi['ppi'].astype(str).str.replace('\xa0', '')
aktobe_ppi['ppi'] = aktobe_ppi['ppi'].astype(str).str.replace('*', '')
aktobe_ppi['ppi'] = aktobe_ppi['ppi'].astype(float)

aktobe_ppi
```

```python
# Suppose the column with the values is the second column in the dataframe
col_to_check = almaty_city_employment.columns[2]  # adjust index if needed

# Get the values for the two rows
value_auezov_2019 = almaty_city_employment.loc[
    (almaty_city_employment['rayon'] == 'Auezov') & (almaty_city_employment['year'] == 2019),
    col_to_check
].values

value_medeu_2021 = almaty_city_employment.loc[
    (almaty_city_employment['rayon'] == 'Medeu') & (almaty_city_employment['year'] == 2021),
    col_to_check
].values

# Check if both values match
if (len(value_auezov_2019) > 0 and value_auezov_2019[0] == 138.282) and \
   (len(value_medeu_2021) > 0 and value_medeu_2021[0] == 115.246):
    print("✅ The statement is TRUE")
else:
    print("❌ The statement is FALSE")
```

## PPI

```python
# 1. Read PPI data
alm_city_ppi_raw = pd.read_excel(
    Path("../Data/PPI/Almaty-city.xls"),
    header=3,
    usecols='X:AH',
    nrows=8,
    na_values=['x', 'х', '-', '', ' ', '...']
)

# 2. Select the needed row (4th from bottom)
alm_city_ppi_raw = alm_city_ppi_raw.iloc[[-4]]

# 3. Assign year columns
alm_city_ppi_raw.columns = [
    2014, 2015, 2016, 2017, 2018,
    2019, 2020, 2021, 2022, 2023, 2024
]

# 4. Extract PPI values correctly (1D array)
ppi_values = alm_city_ppi_raw.iloc[0].values

# 5. Get rayons
rayons = alm_city_inv['rayon'].unique()

# 6. Build long-format dataframe
dfs = []
for rayon in rayons:
    df_temp = pd.DataFrame({
        'rayon': rayon,
        'year': range(2014, 2025),
        'ppi': ppi_values
    })
    dfs.append(df_temp)

alm_city_ppi = pd.concat(dfs, ignore_index=True)

# 7. Clean PPI values
alm_city_ppi['ppi'] = (
    alm_city_ppi['ppi']
    .astype(str)
    .str.replace(',', '.', regex=False)
    .str.replace(')', '', regex=False)
    .str.replace(' ', '', regex=False)
    .str.replace('\xa0', '', regex=False)
    .str.replace('*', '', regex=False)
    .astype(float)
)

# 8. Result
alm_city_ppi
```

```python
# ----- ALL investments -----
akmola_all_2014['rayon'] = akmola_all_2014['rayon'].replace(rayon_map)
akmola_all_2016['rayon'] = akmola_all_2016['rayon'].replace(rayon_map)
akmola_all_2017['rayon'] = akmola_all_2017['rayon'].replace(rayon_map)
akmola_all_2018['rayon'] = akmola_all_2018['rayon'].replace(rayon_map)
akmola_all_2019['rayon'] = akmola_all_2019['rayon'].replace(rayon_map)
akmola_all_2020['rayon'] = akmola_all_2020['rayon'].replace(rayon_map)
akmola_all_2021['rayon'] = akmola_all_2021['rayon'].replace(rayon_map)
akmola_all_2022['rayon'] = akmola_all_2022['rayon'].replace(rayon_map)
akmola_all_2023['rayon'] = akmola_all_2023['rayon'].replace(rayon_map)
akmola_all_2024['rayon'] = akmola_all_2024['rayon'].replace(rayon_map)

# ----- NON-extractive investments -----
akmola_non_2014['rayon'] = akmola_non_2014['rayon'].replace(rayon_map)
akmola_non_2016['rayon'] = akmola_non_2016['rayon'].replace(rayon_map)
akmola_non_2017['rayon'] = akmola_non_2017['rayon'].replace(rayon_map)
akmola_non_2018['rayon'] = akmola_non_2018['rayon'].replace(rayon_map)
akmola_non_2019['rayon'] = akmola_non_2019['rayon'].replace(rayon_map)
akmola_non_2020['rayon'] = akmola_non_2020['rayon'].replace(rayon_map)
akmola_non_2021['rayon'] = akmola_non_2021['rayon'].replace(rayon_map)
akmola_non_2022['rayon'] = akmola_non_2022['rayon'].replace(rayon_map)
akmola_non_2023['rayon'] = akmola_non_2023['rayon'].replace(rayon_map)
akmola_non_2024['rayon'] = akmola_non_2024['rayon'].replace(rayon_map)
```

```python
akmola_2016 = akmola_all_2016.merge(akmola_non_2016, on='rayon', how='left')
akmola_2017 = akmola_all_2017.merge(akmola_non_2017, on='rayon', how='left')
akmola_2018 = akmola_all_2018.merge(akmola_non_2018, on='rayon', how='left')
akmola_2019 = akmola_all_2019.merge(akmola_non_2019, on='rayon', how='left')
akmola_2020 = akmola_all_2020.merge(akmola_non_2020, on='rayon', how='left')
akmola_2021 = akmola_all_2021.merge(akmola_non_2021, on='rayon', how='left')
akmola_2022 = akmola_all_2022.merge(akmola_non_2022, on='rayon', how='left')
akmola_2023 = akmola_all_2023.merge(akmola_non_2023, on='rayon', how='left')
akmola_2024 = akmola_all_2024.merge(akmola_non_2024, on='rayon', how='left')

dfs = [akmola_2016, akmola_2017, akmola_2018, akmola_2019, akmola_2020, akmola_2021, akmola_2022, akmola_2023, akmola_2024]

for i in range(9):
    df = dfs[i]
    df['oblast'] = 'Akmola'
    df['year'] = 2016 + i

akmola_inv = pd.concat([akmola_2016, akmola_2017, akmola_2018, akmola_2019, akmola_2020, akmola_2021, akmola_2022, akmola_2023, akmola_2024], ignore_index=True)
akmola_inv['rayon'] = (
    akmola_inv['rayon']
        .str.strip()          # remove hidden spaces
        .replace(rayon_map)   # enforce English names
)
akmola_inv
```

```python
kokshetau_city = pd.read_excel(Path("../Data/Controls/Akmola_All.xlsx"), sheet_name="г.а. Кокшетау", header=2, usecols='A, Y:AH', nrows=149, na_values=['x', 'х', '-', '', ' ', '...'])

kokshetau_city.columns = ['Indicator', 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
kokshetau_city = kokshetau_city.loc[[49, 81, 133]].reset_index(drop=True)
kokshetau_city['Indicator'] = ['production', 'agriculture', 'retail']
for i in [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]:
    kokshetau_city[i] = kokshetau_city[i].astype(str)
    kokshetau_city[i] = kokshetau_city[i].str.replace(',', '.')
    kokshetau_city[i] = kokshetau_city[i].str.replace(')', '')
    kokshetau_city[i] = kokshetau_city[i].str.replace(' ', '')
    kokshetau_city[i] = kokshetau_city[i].str.replace('\xa0', '')
    kokshetau_city[i] = kokshetau_city[i].str.replace('*', '')
    kokshetau_city[i] = kokshetau_city[i].astype(float)

kokshetau_city = pd.melt(kokshetau_city, id_vars=['Indicator'], var_name='year', value_name='value')
kokshetau_city = kokshetau_city.pivot(index='year', columns='Indicator', values='value').reset_index()
kokshetau_city.insert(0, 'rayon', 'Kokshetau city')
kokshetau_city.columns.name = None
kokshetau_city
```

## PPI

```python
akmola_ppi = pd.read_excel(Path("../Data/PPI/Akmola.xls"), header=2, usecols='Y:AH', nrows=216, na_values=['x', 'х', '-', '', ' ', '...'])

akmola_ppi.columns = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]

rayons = list(akmola_inv['rayon'].unique())

dfs = []

for rayon in rayons:
    df_temp = pd.DataFrame({
        'rayon': rayon,
        'year': range(2014, 2024),
        'ppi': akmola_ppi.values[69]
    })
    dfs.append(df_temp)

akmola_ppi = pd.concat(dfs, ignore_index=True)

akmola_ppi['ppi'] = akmola_ppi['ppi'].astype(str).str.replace(',', '.')
akmola_ppi['ppi'] = akmola_ppi['ppi'].astype(str).str.replace(')', '')
akmola_ppi['ppi'] = akmola_ppi['ppi'].astype(str).str.replace(' ', '')
akmola_ppi['ppi'] = akmola_ppi['ppi'].astype(str).str.replace('\xa0', '')
akmola_ppi['ppi'] = akmola_ppi['ppi'].astype(str).str.replace('*', '')
akmola_ppi['ppi'] = akmola_ppi['ppi'].astype(float)

akmola_ppi
```

## Combine

```python
dfs_akmola = [akmola_inv, akmola_employment, akmola_controls, akmola_ppi]
akmola_merged = reduce(lambda left, right: pd.merge(left, right, on=['rayon', 'year'], how='outer'), dfs_akmola)
akmola_merged
```

```python
rayon_map_astana = {

    # ---- City-level names ----
    'Астана қ.': 'Astana city',
    'Астана қаласы': 'Astana city',
    'Нұр-Сұлтан қаласы': 'Astana city',
    'г. Астана': 'Astana city',

    # ---- Almaty ----
    '"Алматы" ауданы': 'Almaty',
    'Алматы ауданы': 'Almaty',
    'район Алматы ': 'Almaty',
    'район Алматы': 'Almaty',

    # ---- Yesil ----
    '"Есіл" ауданы': 'Yesil',
    'Есіл ауданы': 'Yesil',
    'район Есиль': 'Yesil',
    'район Есиль ': 'Yesil',

    # ---- Saryarka ----
    '"Сарыарқа" ауданы': 'Saryarka',
    'Сарыарқа ауданы': 'Saryarka',
    'район Сарыарка ': 'Saryarka',
    'район Сарыарка': 'Saryarka',

    # ---- Baikonur ----
    '"Байқоңыр" ауданы': 'Baikonur',
    'Байқоңыр ауданы': 'Baikonur',
    'район Байконыр ': 'Baikonur',
    'район Байкоңыр ': 'Baikonur',
    'район Байқоңыр': 'Baikonur',

    # ---- Nura ----
    'район Нура ': 'Nura',
    'район Нұра ': 'Nura'
}
```

```python
# ----- ALL investments -----
astana_city_all_2014['rayon'] = astana_city_all_2014['rayon'].replace(rayon_map_astana)
astana_city_all_2015['rayon'] = astana_city_all_2015['rayon'].replace(rayon_map_astana)
astana_city_all_2016['rayon'] = astana_city_all_2016['rayon'].replace(rayon_map_astana)
astana_city_all_2017['rayon'] = astana_city_all_2017['rayon'].replace(rayon_map_astana)
astana_city_all_2018['rayon'] = astana_city_all_2018['rayon'].replace(rayon_map_astana)
astana_city_all_2019['rayon'] = astana_city_all_2019['rayon'].replace(rayon_map_astana)
astana_city_all_2020['rayon'] = astana_city_all_2020['rayon'].replace(rayon_map_astana)
astana_city_all_2021['rayon'] = astana_city_all_2021['rayon'].replace(rayon_map_astana)
astana_city_all_2022['rayon'] = astana_city_all_2022['rayon'].replace(rayon_map_astana)
astana_city_all_2023['rayon'] = astana_city_all_2023['rayon'].replace(rayon_map_astana)
astana_city_all_2024['rayon'] = astana_city_all_2024['rayon'].replace(rayon_map_astana)

# ----- NON-extractive investments -----
astana_city_non_2015['rayon'] = astana_city_non_2015['rayon'].replace(rayon_map_astana)
astana_city_non_2016['rayon'] = astana_city_non_2016['rayon'].replace(rayon_map_astana)
astana_city_non_2017['rayon'] = astana_city_non_2017['rayon'].replace(rayon_map_astana)
astana_city_non_2018['rayon'] = astana_city_non_2018['rayon'].replace(rayon_map_astana)
astana_city_non_2019['rayon'] = astana_city_non_2019['rayon'].replace(rayon_map_astana)
astana_city_non_2020['rayon'] = astana_city_non_2020['rayon'].replace(rayon_map_astana)
astana_city_non_2021['rayon'] = astana_city_non_2021['rayon'].replace(rayon_map_astana)
astana_city_non_2022['rayon'] = astana_city_non_2022['rayon'].replace(rayon_map_astana)
astana_city_non_2023['rayon'] = astana_city_non_2023['rayon'].replace(rayon_map_astana)
astana_city_non_2024['rayon'] = astana_city_non_2024['rayon'].replace(rayon_map_astana)
```

```python
astana_city_2015 = astana_city_all_2015.merge(astana_city_non_2015, on='rayon', how='left')
astana_city_2016 = astana_city_all_2016.merge(astana_city_non_2016, on='rayon', how='left')
astana_city_2017 = astana_city_all_2017.merge(astana_city_non_2017, on='rayon', how='left')
astana_city_2018 = astana_city_all_2018.merge(astana_city_non_2018, on='rayon', how='left')
astana_city_2019 = astana_city_all_2019.merge(astana_city_non_2019, on='rayon', how='left')
astana_city_2020 = astana_city_all_2020.merge(astana_city_non_2020, on='rayon', how='left')
astana_city_2021 = astana_city_all_2021.merge(astana_city_non_2021, on='rayon', how='left')
astana_city_2022 = astana_city_all_2022.merge(astana_city_non_2022, on='rayon', how='left')
astana_city_2023 = astana_city_all_2023.merge(astana_city_non_2023, on='rayon', how='left')
astana_city_2024 = astana_city_all_2024.merge(astana_city_non_2024, on='rayon', how='left')

dfs = [astana_city_2015, astana_city_2016, astana_city_2017, astana_city_2018, astana_city_2019, astana_city_2020, astana_city_2021, astana_city_2022, astana_city_2023, astana_city_2024]

for i in range(10):
    df = dfs[i]
    df['oblast'] = 'Astana city'
    df['year'] = 2015 + i

astana_city_inv = pd.concat([astana_city_2015, astana_city_2016, astana_city_2017, astana_city_2018, astana_city_2019, astana_city_2020, astana_city_2021, astana_city_2022, astana_city_2023, astana_city_2024], ignore_index=True)
astana_city_inv['rayon'] = (
    astana_city_inv['rayon']
        .str.strip()          # remove hidden spaces
        .replace(rayon_map)   # enforce English names
)
astana_city_inv
```

```python
# Replace Nura with Yesil
astana_city_inv['rayon'] = astana_city_inv['rayon'].replace('Nura', 'Yesil')

# Drop Astana city rows
astana_city_inv = astana_city_inv[astana_city_inv['rayon'] != 'Astana city']

# Group and aggregate
astana_city_inv = (
    astana_city_inv
    .groupby(['year','rayon'], as_index=False)
    .agg({
        'l_all_inv': 'sum',
        'f_all_inv': 'sum',
        'f_nonextr_inv': 'sum',
        'l_nonextr_inv': 'sum'
    })
)

# Add oblast column
astana_city_inv['oblast'] = 'Astana city'
```

## Controls

```python
almaty_rayon = pd.read_excel(Path("../Data/Controls/Astana-city_All.xlsx"), sheet_name="Алматы", header=2, usecols='A, S:AB', nrows=149, na_values=['x', 'х', '-', '', ' ', '...'])

almaty_rayon.columns = ['Indicator', 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
almaty_rayon = almaty_rayon.loc[[82, 57, 141]].reset_index(drop=True)
almaty_rayon['Indicator'] = ['production', 'agriculture', 'retail']
for i in [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]:
    almaty_rayon[i] = almaty_rayon[i].astype(str)
    almaty_rayon[i] = almaty_rayon[i].str.replace(',', '.')
    almaty_rayon[i] = almaty_rayon[i].str.replace(')', '')
    almaty_rayon[i] = almaty_rayon[i].str.replace(' ', '')
    almaty_rayon[i] = almaty_rayon[i].str.replace('\xa0', '')
    almaty_rayon[i] = almaty_rayon[i].str.replace('*', '')
    almaty_rayon[i] = almaty_rayon[i].astype(float)

almaty_rayon = pd.melt(almaty_rayon, id_vars=['Indicator'], var_name='year', value_name='value')
almaty_rayon = almaty_rayon.pivot(index='year', columns='Indicator', values='value').reset_index()
almaty_rayon.insert(0, 'rayon', 'Almaty')
almaty_rayon.columns.name = None
almaty_rayon
```

```python
baikonyr = pd.read_excel(Path("../Data/Controls/Astana-city_All.xlsx"), sheet_name="Байконыр", header=2, usecols='B:H', nrows=149, na_values=['x', 'х', '-', '', ' ', '...'])
baikonyr.columns = ['Indicator', 2018, 2019, 2020, 2021, 2022, 2023]
baikonyr = baikonyr.loc[[82, 57, 141]].reset_index(drop=True)
baikonyr['Indicator'] = ['production', 'agriculture', 'retail']
for i in [2018, 2019, 2020, 2021, 2022, 2023]:
    baikonyr[i] = baikonyr[i].astype(str)
    baikonyr[i] = baikonyr[i].str.replace(',', '.')
    baikonyr[i] = baikonyr[i].str.replace(')', '')
    baikonyr[i] = baikonyr[i].str.replace(' ', '')
    baikonyr[i] = baikonyr[i].str.replace('\xa0', '')
    baikonyr[i] = baikonyr[i].str.replace('*', '')
    baikonyr[i] = baikonyr[i].astype(float)
baikonyr
baikonyr = pd.melt(baikonyr, id_vars=['Indicator'], var_name='year', value_name='value')
baikonyr = baikonyr.pivot(index='year', columns='Indicator', values='value').reset_index()
baikonyr.insert(0, 'rayon', 'Baikonur')
baikonyr.columns.name = None
baikonyr
```

```python
yesil = pd.read_excel(Path("../Data/Controls/Astana-city_All.xlsx"), sheet_name="Есиль", header=2, usecols='B, H:R', nrows=149, na_values=['x', 'х', '-', '', ' ', '...'])
yesil
yesil.columns = ['Indicator', 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
yesil = yesil.loc[[82, 57, 141]].reset_index(drop=True)
yesil['Indicator'] = ['production', 'agriculture', 'retail']
for i in [2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]:
    yesil[i] = yesil[i].astype(str)
    yesil[i] = yesil[i].str.replace(',', '.')
    yesil[i] = yesil[i].str.replace(')', '')
    yesil[i] = yesil[i].str.replace(' ', '')
    yesil[i] = yesil[i].str.replace('\xa0', '')
    yesil[i] = yesil[i].str.replace('*', '')
    yesil[i] = yesil[i].astype(float)

yesil = pd.melt(yesil, id_vars=['Indicator'], var_name='year', value_name='value')
yesil = yesil.pivot(index='year', columns='Indicator', values='value').reset_index()
yesil.insert(0, 'rayon', 'Yesil')
yesil.columns.name = None
yesil
```

```python
saryarka = pd.read_excel(Path("../Data/Controls/Astana-city_All.xlsx"), sheet_name="Сарыарка", header=2, usecols='B, S:AB', nrows=149, na_values=['x', 'х', '-', '', ' ', '...'])
saryarka.columns = ['Indicator', 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
saryarka = saryarka.loc[[82, 57, 141]].reset_index(drop=True)
saryarka['Indicator'] = ['production', 'agriculture', 'retail']
saryarka
for i in [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]:
    saryarka[i] = saryarka[i].astype(str)
    saryarka[i] = saryarka[i].str.replace(',', '.')
    saryarka[i] = saryarka[i].str.replace(')', '')
    saryarka[i] = saryarka[i].str.replace(' ', '')
    saryarka[i] = saryarka[i].str.replace('\xa0', '')
    saryarka[i] = saryarka[i].str.replace('*', '')
    saryarka[i] = saryarka[i].astype(float)

saryarka = pd.melt(saryarka, id_vars=['Indicator'], var_name='year', value_name='value')
saryarka = saryarka.pivot(index='year', columns='Indicator', values='value').reset_index()
saryarka.insert(0, 'rayon', 'Saryarka')
saryarka.columns.name = None
saryarka
```

```python
# Replace Nura with Yesil
astana_city_controls['rayon'] = astana_city_controls['rayon'].replace('Nura', 'Yesil')

# Drop Astana city rows
astana_city_controls = astana_city_controls[astana_city_controls['rayon'] != 'Astana city']

# Group and aggregate
astana_city_controls = (
    astana_city_controls
    .groupby(['year','rayon'], as_index=False)
    .agg({
        'agriculture': 'sum',
        'production': 'sum',
        'retail': 'sum'
    })
)

# Add oblast column
astana_city_controls
```

## PPI

```python
astana_city_ppi = pd.read_excel(
    Path("../Data/PPI/Astana-city.xls"),
    header=2,
    usecols='X:AH',
    nrows=216,
    na_values=['x', 'х', '-', '', ' ', '...']
)

astana_city_ppi.columns = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]

rayons = list(astana_city_inv['rayon'].unique())

dfs = []

for rayon in rayons:
    df_temp = pd.DataFrame({
        'rayon': [rayon] * len(range(2014, 2025)),
        'year': list(range(2014, 2025)),
        'ppi': astana_city_ppi.iloc[1].values
    })
    dfs.append(df_temp)

astana_city_ppi = pd.concat(dfs, ignore_index=True)

astana_city_ppi['ppi'] = astana_city_ppi['ppi'].astype(str).str.replace(',', '.')
astana_city_ppi['ppi'] = astana_city_ppi['ppi'].astype(str).str.replace(')', '')
astana_city_ppi['ppi'] = astana_city_ppi['ppi'].astype(str).str.replace(' ', '')
astana_city_ppi['ppi'] = astana_city_ppi['ppi'].astype(str).str.replace('\xa0', '')
astana_city_ppi['ppi'] = astana_city_ppi['ppi'].astype(str).str.replace('*', '')
astana_city_ppi['ppi'] = astana_city_ppi['ppi'].astype(float)


astana_city_ppi
```

## Combine

```python
dfs_astana_city = [astana_city_inv, astana_city_employment, astana_city_controls, astana_city_ppi]
astana_city_merged = reduce(lambda left, right: pd.merge(left, right, on=['rayon', 'year'], how='outer'), dfs_astana_city)
# Drop Astana city rows
astana_city_merged = astana_city_merged[astana_city_merged['rayon'] != 'Astana city']
astana_city_merged
```

## Other Controls

### Kostanay City

```python
kostanay_city = pd.read_excel(Path("../Data/Controls/Kostanay_All.xlsx"), sheet_name="г. Костанай", header=2, usecols='A, Y:AH', nrows=160, na_values=['x', 'х', '-', '', ' ', '...'])
kostanay_city.columns = ['Indicator', 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
kostanay_city = kostanay_city.loc[[79, 108, 151]].reset_index(drop=True)
kostanay_city['Indicator'] = ['production', 'agriculture', 'retail']
for i in [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]:
    kostanay_city[i] = kostanay_city[i].astype(str)
    kostanay_city[i] = kostanay_city[i].str.replace(',', '.')
    kostanay_city[i] = kostanay_city[i].str.replace(')', '')
    kostanay_city[i] = kostanay_city[i].str.replace(' ', '')
    kostanay_city[i] = kostanay_city[i].str.replace('\xa0', '')
    kostanay_city[i] = kostanay_city[i].str.replace('*', '')
    kostanay_city[i] = kostanay_city[i].astype(float)
kostanay_city = pd.melt(kostanay_city, id_vars=['Indicator'], var_name='year', value_name='value')
kostanay_city = kostanay_city.pivot(index='year', columns='Indicator', values='value').reset_index()
kostanay_city.insert(0, 'rayon', 'Kostanay city')
kostanay_city.columns.name = None
kostanay_city
```

## PPI

```python
kyzylorda_ppi = pd.read_excel(
    Path("../Data/PPI/Kyzylorda.xls"),
    header=2,
    usecols='Y:AH',
    nrows=79,
    na_values=['x', 'х', '-', '', ' ', '...']
)
kyzylorda_ppi.columns = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
rayons = list(kyzylorda_inv['rayon'].unique())

dfs = []

for rayon in rayons:
    df_temp = pd.DataFrame({
        'rayon': [rayon] * len(range(2014, 2024)),
        'year': list(range(2014, 2024)),
        'ppi': kyzylorda_ppi.iloc[-1].values
    })
    dfs.append(df_temp)

kyzylorda_ppi = pd.concat(dfs, ignore_index=True)
kyzylorda_ppi['ppi'] = kyzylorda_ppi['ppi'].astype(str).str.replace(',', '.')
kyzylorda_ppi['ppi'] = kyzylorda_ppi['ppi'].astype(str).str.replace(')', '')
kyzylorda_ppi['ppi'] = kyzylorda_ppi['ppi'].astype(str).str.replace(' ', '')
kyzylorda_ppi['ppi'] = kyzylorda_ppi['ppi'].astype(str).str.replace('\xa0', '')
kyzylorda_ppi['ppi'] = kyzylorda_ppi['ppi'].astype(str).str.replace('*', '')
kyzylorda_ppi['ppi'] = kyzylorda_ppi['ppi'].astype(float)

kyzylorda_ppi
```

## Combine

```python
dfs_kyzylorda = [kyzylorda_inv, kyzylorda_employment, kyzylorda_controls, kyzylorda_ppi]
kyzylorda_merged = reduce(lambda left, right: pd.merge(left, right, on=['rayon', 'year'], how='outer'), dfs_kyzylorda)
kyzylorda_merged
```

## PPI

```python
mangystau_ppi = pd.read_excel(
    Path("../Data/PPI/Mangystau.xls"),
    header=2,
    usecols='Y:AH',
    nrows=65,
    na_values=['x', 'х', '-', '', ' ', '...']
)
mangystau_ppi.columns = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
rayons = list(mangystau_inv['rayon'].unique())

dfs = []

for rayon in rayons:
    df_temp = pd.DataFrame({
        'rayon': [rayon] * len(range(2014, 2024)),
        'year': list(range(2014, 2024)),
        'ppi': mangystau_ppi.iloc[-1].values
    })
    dfs.append(df_temp)

mangystau_ppi = pd.concat(dfs, ignore_index=True)
mangystau_ppi['ppi'] = mangystau_ppi['ppi'].astype(str).str.replace(',', '.')
mangystau_ppi['ppi'] = mangystau_ppi['ppi'].astype(str).str.replace(')', '')
mangystau_ppi['ppi'] = mangystau_ppi['ppi'].astype(str).str.replace(' ', '')
mangystau_ppi['ppi'] = mangystau_ppi['ppi'].astype(str).str.replace('\xa0', '')
mangystau_ppi['ppi'] = mangystau_ppi['ppi'].astype(str).str.replace('*', '')
mangystau_ppi['ppi'] = mangystau_ppi['ppi'].astype(float)

mangystau_ppi
```

```python
petropavl_city = pd.read_excel(Path("../Data/Controls/North-Kazakhstan_All.xlsx"), sheet_name="г.Петропавловск", header=2, usecols='B, Y:AI', nrows=180, na_values=['x', 'х', '-', '', ' ', '...'])
petropavl_city.columns = ['Indicator', 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
petropavl_city = petropavl_city.loc[[85, 113, 156]].reset_index(drop=True)
petropavl_city['Indicator'] = ['production', 'agriculture', 'retail']
for i in [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]:
    petropavl_city[i] = petropavl_city[i].astype(str)
    petropavl_city[i] = petropavl_city[i].str.replace(',', '.')
    petropavl_city[i] = petropavl_city[i].str.replace(')', '')
    petropavl_city[i] = petropavl_city[i].str.replace(' ', '')
    petropavl_city[i] = petropavl_city[i].str.replace('\xa0', '')
    petropavl_city[i] = petropavl_city[i].str.replace('*', '')
    petropavl_city[i] = petropavl_city[i].astype(float)
petropavl_city = pd.melt(petropavl_city, id_vars=['Indicator'], var_name='year', value_name='value')
petropavl_city = petropavl_city.pivot(index='year', columns='Indicator', values='value').reset_index()
petropavl_city.insert(0, 'rayon', 'Petropavl city')
petropavl_city.columns.name = None
petropavl_city
```

## PPI

```python
north_ppi = pd.read_excel(
    Path("../Data/PPI/North-Kazakhstan.xlsx"),
    header=2,
    usecols='Y:AI',
    nrows=199,
    na_values=['x', 'х', '-', '', ' ', '...']
)
north_ppi.columns = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
years = north_ppi.columns.tolist()
rayons = list(north_inv['rayon'].unique())

dfs = []

for rayon in rayons:
    df_temp = pd.DataFrame({
        'rayon': [rayon] * len(years),
        'year': years,
        'ppi': north_ppi.iloc[-1].values
    })
    dfs.append(df_temp)
dfs
north_ppi_long = pd.concat(dfs, ignore_index=True)
north_ppi_long
north_ppi = north_ppi_long
north_ppi['ppi'] = north_ppi['ppi'].astype(str).str.replace(',', '.')
north_ppi['ppi'] = north_ppi['ppi'].astype(str).str.replace(')', '')
north_ppi['ppi'] = north_ppi['ppi'].astype(str).str.replace(' ', '')
north_ppi['ppi'] = north_ppi['ppi'].astype(str).str.replace('\xa0', '')
north_ppi['ppi'] = north_ppi['ppi'].astype(str).str.replace('*', '')
north_ppi['ppi'] = north_ppi['ppi'].astype(float)

north_ppi
```

## Combine

```python
dfs_north = [north_inv, north_employment, north_controls, north_ppi]
north_merged = reduce(lambda left, right: pd.merge(left, right, on=['rayon', 'year'], how='outer'), dfs_north)
north_merged['oblast'] = north_merged['oblast'].fillna('North-Kazakhstan')
north_merged
```

```python
pavlodar_ppi = pd.read_excel(
    Path("../Data/PPI/Pavlodar.xlsx"),
    header=2,
    usecols='Y:AI',
    nrows=83,
    na_values=['x', 'х', '-', '', ' ', '...']
)
pavlodar_ppi.columns = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
rayons = list(pavlodar_inv['rayon'].unique())

dfs = []

for rayon in rayons:
    df_temp = pd.DataFrame({
        'rayon': [rayon] * len(range(2014, 2025)),
        'year': list(range(2014, 2025)),
        'ppi': pavlodar_ppi.iloc[-1].values
    })
    dfs.append(df_temp)

pavlodar_ppi = pd.concat(dfs, ignore_index=True)
pavlodar_ppi['ppi'] = pavlodar_ppi['ppi'].astype(str).str.replace(',', '.')
pavlodar_ppi['ppi'] = pavlodar_ppi['ppi'].astype(str).str.replace(')', '')
pavlodar_ppi['ppi'] = pavlodar_ppi['ppi'].astype(str).str.replace(' ', '')
pavlodar_ppi['ppi'] = pavlodar_ppi['ppi'].astype(str).str.replace('\xa0', '')
pavlodar_ppi['ppi'] = pavlodar_ppi['ppi'].astype(str).str.replace('*', '')
pavlodar_ppi['ppi'] = pavlodar_ppi['ppi'].astype(float)

pavlodar_ppi
```

```python
abay = pd.read_excel(Path("../Data/Controls/Shymkent-city_All.xls"), sheet_name="Абайский район", header=2, usecols='B:I', nrows=180, na_values=['x', 'х', '-', '', ' ', '...'])
abay.columns = ['Indicator', 2017, 2018, 2019, 2020, 2021, 2022, 2023]
abay = abay.loc[[65, 79, 106]].reset_index(drop=True)
abay['Indicator'] = ['production', 'agriculture', 'retail']
for i in [2017, 2018, 2019, 2020, 2021, 2022, 2023]:
    abay[i] = abay[i].astype(str)
    abay[i] = abay[i].str.replace(',', '.')
    abay[i] = abay[i].str.replace(')', '')
    abay[i] = abay[i].str.replace(' ', '')
    abay[i] = abay[i].str.replace('\xa0', '')
    abay[i] = abay[i].str.replace('*', '')
    abay[i] = abay[i].astype(float)
abay = pd.melt(abay, id_vars=['Indicator'], var_name='year', value_name='value')
abay = abay.pivot(index='year', columns='Indicator', values='value').reset_index()
abay.insert(0, 'rayon', 'Abay')
abay.columns.name = None
abay
```

## PPI

```python
shymkent_city_ppi = pd.read_excel(
    Path("../Data/PPI/Shymkent-city.xls"),
    header=2,
    usecols='Y:AH',
    nrows=54,
    na_values=['x', 'х', '-', '', ' ', '...']
)
shymkent_city_ppi.columns = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
rayons = list(shymkent_city_inv['rayon'].unique())

dfs = []

for rayon in rayons:
    df_temp = pd.DataFrame({
        'rayon': [rayon] * len(range(2014, 2024)),
        'year': list(range(2014, 2024)),
        'ppi': shymkent_city_ppi.iloc[-1].values
    })
    dfs.append(df_temp)
shymkent_city_ppi = pd.concat(dfs, ignore_index=True)
shymkent_city_ppi['ppi'] = shymkent_city_ppi['ppi'].astype(str).str.replace(',', '.')
shymkent_city_ppi['ppi'] = shymkent_city_ppi['ppi'].astype(str).str.replace(')', '')
shymkent_city_ppi['ppi'] = shymkent_city_ppi['ppi'].astype(str).str.replace(' ', '')
shymkent_city_ppi['ppi'] = shymkent_city_ppi['ppi'].astype(str).str.replace('\xa0', '')
shymkent_city_ppi['ppi'] = shymkent_city_ppi['ppi'].astype(str).str.replace('*', '')
shymkent_city_ppi['ppi'] = shymkent_city_ppi['ppi'].astype(float)

shymkent_city_ppi.loc[
    shymkent_city_ppi['year'].isin([2014, 2015, 2016, 2017]),
    'ppi'
] = ''

shymkent_city_ppi
```

```python
# ----- ALL investments -----
turkestan_all_2014['rayon'] = turkestan_all_2014['rayon'].replace(rayon_map_turkestan)
turkestan_all_2015['rayon'] = turkestan_all_2015['rayon'].replace(rayon_map_turkestan)
turkestan_all_2016['rayon'] = turkestan_all_2016['rayon'].replace(rayon_map_turkestan)
turkestan_all_2017['rayon'] = turkestan_all_2017['rayon'].replace(rayon_map_turkestan)
turkestan_all_2018['rayon'] = turkestan_all_2018['rayon'].replace(rayon_map_turkestan)
turkestan_all_2019['rayon'] = turkestan_all_2019['rayon'].replace(rayon_map_turkestan)
turkestan_all_2020['rayon'] = turkestan_all_2020['rayon'].replace(rayon_map_turkestan)
turkestan_all_2021['rayon'] = turkestan_all_2021['rayon'].replace(rayon_map_turkestan)
turkestan_all_2022['rayon'] = turkestan_all_2022['rayon'].replace(rayon_map_turkestan)
turkestan_all_2023['rayon'] = turkestan_all_2023['rayon'].replace(rayon_map_turkestan)
turkestan_all_2024['rayon'] = turkestan_all_2024['rayon'].replace(rayon_map_turkestan)

# ----- NON-extractive investments -----
turkestan_non_2014['rayon'] = turkestan_non_2014['rayon'].replace(rayon_map_turkestan)
turkestan_non_2015['rayon'] = turkestan_non_2015['rayon'].replace(rayon_map_turkestan)
turkestan_non_2016['rayon'] = turkestan_non_2016['rayon'].replace(rayon_map_turkestan)
turkestan_non_2017['rayon'] = turkestan_non_2017['rayon'].replace(rayon_map_turkestan)
turkestan_non_2018['rayon'] = turkestan_non_2018['rayon'].replace(rayon_map_turkestan)
turkestan_non_2019['rayon'] = turkestan_non_2019['rayon'].replace(rayon_map_turkestan)
turkestan_non_2020['rayon'] = turkestan_non_2020['rayon'].replace(rayon_map_turkestan)
turkestan_non_2021['rayon'] = turkestan_non_2021['rayon'].replace(rayon_map_turkestan)
turkestan_non_2022['rayon'] = turkestan_non_2022['rayon'].replace(rayon_map_turkestan)
turkestan_non_2023['rayon'] = turkestan_non_2023['rayon'].replace(rayon_map_turkestan)
turkestan_non_2024['rayon'] = turkestan_non_2024['rayon'].replace(rayon_map_turkestan)
```

## PPI

```python
turkestan_ppi = pd.read_excel(
    Path("../Data/PPI/Turkistan.xls"),
    header=2,
    usecols='Y:AI',
    nrows=64,
    na_values=['x', 'х', '-', '', ' ', '...']
)
turkestan_ppi.columns = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
rayons = list(turkestan_inv['rayon'].unique())

dfs = []

for rayon in rayons:
    df_temp = pd.DataFrame({
        'rayon': [rayon] * len(range(2014, 2025)),
        'year': list(range(2014, 2025)),
        'ppi': turkestan_ppi.iloc[-1].values
    })
    dfs.append(df_temp)

turkestan_ppi = pd.concat(dfs, ignore_index=True)
turkestan_ppi['ppi'] = turkestan_ppi['ppi'].astype(str).str.replace(',', '.')
turkestan_ppi['ppi'] = turkestan_ppi['ppi'].astype(str).str.replace(')', '')
turkestan_ppi['ppi'] = turkestan_ppi['ppi'].astype(str).str.replace(' ', '')
turkestan_ppi['ppi'] = turkestan_ppi['ppi'].astype(str).str.replace('\xa0', '')
turkestan_ppi['ppi'] = turkestan_ppi['ppi'].astype(str).str.replace('*', '')
turkestan_ppi['ppi'] = turkestan_ppi['ppi'].astype(float)

turkestan_ppi
```

```python
dfs_turkestan = [turkestan_inv, turkestan_employment, turkestan_controls, turkestan_ppi]
turkestan_merged = reduce(lambda left, right: pd.merge(left, right, on=['rayon', 'year'], how='outer'), dfs_turkestan)
turkestan_merged['oblast'] = turkestan_merged['oblast'].fillna('Turkestan')
turkestan_merged
```

```python
Akzhaiyk = pd.read_excel(Path("../Data/Controls/West-Kazakhstan_All.xls"), sheet_name="Акжаик", header=2, usecols='A, Y:AH', nrows=180, na_values=['x', 'х', '-', '', ' ', '...'])
Akzhaiyk.columns = ['Indicator', 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
Akzhaiyk = Akzhaiyk.loc[[68, 86, 130]].reset_index(drop=True)
Akzhaiyk['Indicator'] = ['production', 'agriculture', 'retail']
for i in [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]:
    Akzhaiyk[i] = Akzhaiyk[i].astype(str)
    Akzhaiyk[i] = Akzhaiyk[i].str.replace(',', '.')
    Akzhaiyk[i] = Akzhaiyk[i].str.replace(')', '')
    Akzhaiyk[i] = Akzhaiyk[i].str.replace(' ', '')
    Akzhaiyk[i] = Akzhaiyk[i].str.replace('\xa0', '')
    Akzhaiyk[i] = Akzhaiyk[i].str.replace('*', '')
    Akzhaiyk[i] = Akzhaiyk[i].astype(float)
Akzhaiyk = pd.melt(Akzhaiyk, id_vars=['Indicator'], var_name='year', value_name='value')
Akzhaiyk = Akzhaiyk.pivot(index='year', columns='Indicator', values='value').reset_index()
Akzhaiyk.insert(0, 'rayon', 'Akzhaiyk')
Akzhaiyk.columns.name = None
Akzhaiyk
```

```python
all_columns = ['rayon', 'l_all_inv', 'f_all_inv']
non_columns = ['rayon', 'all_nonextr_inv', 'f_nonextr_inv']

for df in [zhambyl_all_2015, zhambyl_all_2016, zhambyl_all_2017, zhambyl_all_2018, zhambyl_all_2019,
      zhambyl_all_2020, zhambyl_all_2021, zhambyl_all_2022, zhambyl_all_2023, zhambyl_all_2024]:
    df.columns = all_columns

for df in [zhambyl_non_2016, zhambyl_non_2017, zhambyl_non_2018, zhambyl_non_2019, zhambyl_non_2020,
      zhambyl_non_2021, zhambyl_non_2022, zhambyl_non_2023, zhambyl_non_2024]:
    df.columns = non_columns

    df[['all_nonextr_inv', 'f_nonextr_inv']] = (
        df[['all_nonextr_inv', 'f_nonextr_inv']]
        .astype(str)
        .apply(lambda x: x.str.replace(' ', ''))
        .astype(float)
    )

    df['l_nonextr_inv'] = df['all_nonextr_inv'] - df['f_nonextr_inv']
    df.drop('all_nonextr_inv', axis=1, inplace=True)
```

```python
# ----- ALL investments -----
zhambyl_all_2015['rayon'] = zhambyl_all_2015['rayon'].replace(rayon_map_zhambyl)
zhambyl_all_2016['rayon'] = zhambyl_all_2016['rayon'].replace(rayon_map_zhambyl)
zhambyl_all_2017['rayon'] = zhambyl_all_2017['rayon'].replace(rayon_map_zhambyl)
zhambyl_all_2018['rayon'] = zhambyl_all_2018['rayon'].replace(rayon_map_zhambyl)
zhambyl_all_2019['rayon'] = zhambyl_all_2019['rayon'].replace(rayon_map_zhambyl)
zhambyl_all_2020['rayon'] = zhambyl_all_2020['rayon'].replace(rayon_map_zhambyl)
zhambyl_all_2021['rayon'] = zhambyl_all_2021['rayon'].replace(rayon_map_zhambyl)
zhambyl_all_2022['rayon'] = zhambyl_all_2022['rayon'].replace(rayon_map_zhambyl)
zhambyl_all_2023['rayon'] = zhambyl_all_2023['rayon'].replace(rayon_map_zhambyl)
zhambyl_all_2024['rayon'] = zhambyl_all_2024['rayon'].replace(rayon_map_zhambyl)

# ----- NON-extractive investments -----
zhambyl_non_2016['rayon'] = zhambyl_non_2016['rayon'].replace(rayon_map_zhambyl)
zhambyl_non_2017['rayon'] = zhambyl_non_2017['rayon'].replace(rayon_map_zhambyl)
zhambyl_non_2018['rayon'] = zhambyl_non_2018['rayon'].replace(rayon_map_zhambyl)
zhambyl_non_2019['rayon'] = zhambyl_non_2019['rayon'].replace(rayon_map_zhambyl)
zhambyl_non_2020['rayon'] = zhambyl_non_2020['rayon'].replace(rayon_map_zhambyl)
zhambyl_non_2021['rayon'] = zhambyl_non_2021['rayon'].replace(rayon_map_zhambyl)
zhambyl_non_2022['rayon'] = zhambyl_non_2022['rayon'].replace(rayon_map_zhambyl)
zhambyl_non_2023['rayon'] = zhambyl_non_2023['rayon'].replace(rayon_map_zhambyl)
zhambyl_non_2024['rayon'] = zhambyl_non_2024['rayon'].replace(rayon_map_zhambyl)
```

```python
zhambyl_2015 = zhambyl_all_2015
zhambyl_2016 = zhambyl_all_2016.merge(zhambyl_non_2016, on='rayon', how='left')
zhambyl_2017 = zhambyl_all_2017.merge(zhambyl_non_2017, on='rayon', how='left')
zhambyl_2018 = zhambyl_all_2018.merge(zhambyl_non_2018, on='rayon', how='left')
zhambyl_2019 = zhambyl_all_2019.merge(zhambyl_non_2019, on='rayon', how='left')
zhambyl_2020 = zhambyl_all_2020.merge(zhambyl_non_2020, on='rayon', how='left')
zhambyl_2021 = zhambyl_all_2021.merge(zhambyl_non_2021, on='rayon', how='left')
zhambyl_2022 = zhambyl_all_2022.merge(zhambyl_non_2022, on='rayon', how='left')
zhambyl_2023 = zhambyl_all_2023.merge(zhambyl_non_2023, on='rayon', how='left')
zhambyl_2024 = zhambyl_all_2024.merge(zhambyl_non_2024, on='rayon', how='left')

dfs = [zhambyl_2015, zhambyl_2016, zhambyl_2017, zhambyl_2018, zhambyl_2019, zhambyl_2020, zhambyl_2021, zhambyl_2022, zhambyl_2023, zhambyl_2024]

for i in range(10):
    df = dfs[i]
    df['oblast'] = 'Zhambyl'
    df['year'] = 2015 + i

zhambyl_inv = pd.concat([zhambyl_2015, zhambyl_2016, zhambyl_2017, zhambyl_2018, zhambyl_2019, zhambyl_2020, zhambyl_2021, zhambyl_2022, zhambyl_2023, zhambyl_2024], ignore_index=True)
zhambyl_inv['rayon'] = (
    zhambyl_inv['rayon']
        .str.strip()          # remove hidden spaces
        .replace(rayon_map)   # enforce English names
)
zhambyl_inv
```

## PPI

```python
zhambyl_ppi = pd.read_excel(
    Path("../Data/PPI/Zhambyl.xls"),
    header=2,
    usecols='Y:AI',
    nrows=78,
    na_values=['x', 'х', '-', '', ' ', '...']
)
zhambyl_ppi.columns = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
rayons = list(zhambyl_inv['rayon'].unique())
zhambyl_ppi
dfs = []

for rayon in rayons:
    df_temp = pd.DataFrame({
        'rayon': [rayon] * len(range(2014, 2025)),
        'year': list(range(2014, 2025)),
        'ppi': zhambyl_ppi.iloc[-1].values
    })
    dfs.append(df_temp)

zhambyl_ppi = pd.concat(dfs, ignore_index=True)
zhambyl_ppi['ppi'] = zhambyl_ppi['ppi'].astype(str).str.replace(',', '.')
zhambyl_ppi['ppi'] = zhambyl_ppi['ppi'].astype(str).str.replace(')', '')
zhambyl_ppi['ppi'] = zhambyl_ppi['ppi'].astype(str).str.replace(' ', '')
zhambyl_ppi['ppi'] = zhambyl_ppi['ppi'].astype(str).str.replace('\xa0', '')
zhambyl_ppi['ppi'] = zhambyl_ppi['ppi'].astype(str).str.replace('*', '')
zhambyl_ppi['ppi'] = zhambyl_ppi['ppi'].astype(float)

zhambyl_ppi
```

## Combine

```python
dfs_zhambyl = [zhambyl_inv, zhambyl_employment, zhambyl_controls, zhambyl_ppi]
zhambyl_merged = reduce(lambda left, right: pd.merge(left, right, on=['rayon', 'year'], how='outer'), dfs_zhambyl)
zhambyl_merged['oblast'] = zhambyl_merged['oblast'].fillna('Zhambyl')
zhambyl_merged
```

# Combine all oblast

```python
all_merged = pd.concat([east_merged, akmola_merged, aktobe_merged, alm_city_merged, almaty_merged, astana_city_merged, atyrau_merged,
                       karagandy_merged, kostanay_merged, kyzylorda_merged, mangystau_merged, north_merged, pavlodar_merged,
                       shymkent_city_merged, turkestan_merged, west_merged, zhambyl_merged, almaty_merged], ignore_index=True)
all_merged
```
